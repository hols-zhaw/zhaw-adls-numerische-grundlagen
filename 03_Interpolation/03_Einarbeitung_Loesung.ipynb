### Aufgabe 2: Lineare Interpolation


Studiere als Einstieg dieses Video: ["Lineare Interpolation, Herleitung, Formel" - Mathe by Daniel Jung](https://www.youtube.com/watch?v=KYiIGZYrb9M).

**a)** Seien wie im Video folgende Punkte gegeben:

\begin{align*}
(x_1, y_1) &= (2, 4) \\
(x_2, y_2) &= (3, 25)
\end{align*}

wobei $y=f(x)$ mit einer nicht bekannten Funktion $f:\mathbb{R}\to\mathbb{R}$.

Bestimme die lineare Interpolation bei $x=2.4$ mithilfe der Gleichung im Video.


**a) Lösung**

Die Gleichung für die lineare Interpolation (Gerade durch zwei Punkte) lautet:

$$ y = y_1 + \frac{y_2 - y_1}{x_2 - x_1}\cdot\left(x - x_1\right) $$

Setzt man die beiden Punkte und $x=2.4$ ein, erhält man:

\begin{align*}
y &= 4 + \frac{25 - 4}{3 - 2}\cdot\left(2.4 - 2\right) \\
&= 4 + 21 \cdot 0.4 \\
&= 12.4
\end{align*}


**b)** Implementiere die lineare Interpolation zwischen zwei Punkten in Python.

-   Definiere dazu eine Funktion `linear_interpolation(x_data, y_data, x)`, die ein Array von $y$-Werten für die gegebenen $x$-Werte `x` zurückgibt.
-   `x_data` und `y_data` sind Arrays der Länge 2 mit den bekannten Punkten.
-   Teste die Funktion mit den Daten von oben am Punkt $x=2.4$ und vergleiche mit dem Resultat von a).


```python
# Lösung 2.b)


def linear_interpolation(x_data, y_data, x):
    m = (y_data[1] - y_data[0]) / (x_data[1] - x_data[0])
    return m * (x - x_data[0]) + y_data[0]


x_data = np.array([2, 3])
y_data = np.array([4, 25])

linear_interpolation(x_data, y_data, 2.4)
```

**c)** Darstellen

-   Stelle die beiden gegebenen Punkte inkl. einer Verbindungslinie dar, indem du `plt.plot` mit der Option `"-o"` (verbundene Punkte) verwendest.
-   Berechne mit der Funktion aus Teil b) für 9 $x$-Werte im Intervall $[2.1, 2.9]$ die interpolierten $y$-Werte, und füge sie dem Diagramm als rote Punkte hinzu.
-   Deckt sich das Resultat mit deiner Erwartung?


```python
# Lösung 2.c)

x = np.linspace(2.1, 2.9, 9)
y = linear_interpolation(x_data, y_data, x)

plt.plot(x_data, y_data, "-o", label="Daten")
plt.plot(x, y, "ro", label="Interpolation")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Lineare Interpolation")
plt.show()
```

### Aufgabe 4: Kubische Spline-Interpolation

Schau dir folgendes an:

In beiden Videos wird ist die Mathematik etwas zu schnell erklärt. Keine Angst, wenn du das nicht gleich alles nachvollziehen kannst.

Ein kubischer Spline ist eine Interpolationsmethode, die mehrere Polynome verwendet, um die Datenpunkte zu verbinden. Jedes Polynom verbindet zwei benachbarte Datenpunkte (Stützstellen) und ist so aufgebaut:

$$ s_j(x) = a_j + b_j(x - x_j) + c_j(x - x_j)^2 + d_j(x - x_j)^3 \quad\text{für}\quad j=0,1,\ldots,n-2 $$

wobei $x_j$ die Stützstellen und $a_j$, $b_j$, $c_j$ und $d_j$ die Koeffizienten des Polynoms sind. Hier ist $n$ die Anzahl der Stützstellen.

Fragen:

**a)** Wie unterscheidet sich ein kubischer Spline durch mehrere Punkte von einem einzelnen Interpolationspolynom?


**a) Lösung:** Ein kubischer Spline ist eine stückweise definierte Funktion, die in jedem Intervall zwischen zwei benachbarten Punkten ein eigenes kubisches Polynom definiert. Das Interpolationspolynom ist ein einziges Polynom (Grad = Anzahl Punkte + 1), das alle Punkte durchläuft.


**b)** Wenn wir 8 Datenpunkte $(x_i, y_i)$ für $i=0,1,\ldots,7$ gegeben haben, aus wie vielen Polynomen besteht dann der kubische Spline?


**b) Lösung:** Der kubische Spline besteht aus 7 Polynomen, je eines für jedes Intervall zwischen zwei benachbarten Punkten.


**c)** Wie viele Koeffizienten gibt es dann in diesem Fall zu bestimmen?


**c) Lösung:** Für jedes der 7 Polynome gibt es 4 Koeffizienten: $7\cdot 4 = 28$ Koeffizienten.


**d)** Wir nehmen natürliche Randbedingungen an, das heisst, die zweite Ableitung soll an den äusseren Randpunkten ($x_0$ und $x_7$) gleich Null sein. Welche Bedingungen müssen die Polynome erfüllen? Und wie viele Bedingungen von welchem Typ gibt es?


**d) Lösung:**

-   Jedes Polynom muss durch die beiden benachbarten Punkte gehen:
    -   $s_j(x_j) = y_j$ und $s_j(x_{j+1}) = y_{j+1}$ für $j=0,1,\ldots,6$
    -   Das sind $7\cdot 2 = 14$ Bedingungen.
-   Jeweils die erste und die zweite Ableitung der Polynome müssen an den inneren Punkten übereinstimmen.
    -   $s_j'(x_{j+1}) = s_{j+1}'(x_{j+1})$ und $s_j''(x_{j+1}) = s_{j+1}''(x_{j+1})$ für $j=0,1,\ldots,5$ (insgesamt 6 innere Punkte)
    -   Das sind $(8-2)\cdot 2 = 12$ Bedingungen.
-   Bei natürlichen Randbedingungen muss die zweite Ableitung an den Rändern 0 sein
    -   $s_0''(x_0) = 0$ und $s_6''(x_7) = 0$
    -   Das sind 2 Bedingungen.
