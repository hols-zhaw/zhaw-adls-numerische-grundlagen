## Numerische Ableitung ausgehend von einer Funktion

Es sei eine Funktion $f : D \to \mathbb{R}$ mit $D \subseteq \mathbb{R}$ gegeben, ihre Ableitung kennen wir aber nicht. Darunter können wir uns eine Funktion vorstellen, die z.B. das Resultat einer numerischen Simulation ist, oder die aus Messwerten via Interpolation gewonnen wird.

Die Ableitung an einer Stelle $x\in D$ ist definiert als der Grenzwert des Differenzenquotienten

$$ f'(x) = \lim_{h\to 0} \frac{f(x+h) - f(x)}{h} $$

Als numerische Schätzung für $f'(x)$ können wir den Differenzenquotienten für ein kleines $h$ verwenden ($0 < h \ll 1 $), also

$$ f'(x) \simeq \frac{f(x+h) - f(x)}{h} $$

Wie gross $h$ zu wählen ist, muss man ausprobieren, und hängt vom Verhalten der Funktion ab.

Zudem könnten wir statt einen Schritt um $h$ nach vorne zu machen auch einen Schritt zurück machen, so dass ein "rückwärtsgerichteter" Differenzenquotient berechnet wird. Eine gute Idee ist, den Mittelwert zwischen dem vorwärtsgerichteten und dem rückwärtsgerichteten Differenzenquotienten zu nehmen. Dieser wird als **zentraler Differenzenquotient** bezeichnet:

$$ f'(x) \simeq \frac{1}{2}\left( \frac{f(x+h) - f(x)}{h} + \frac{f(x) - f(x-h)}{h} \right) = \frac{f(x+h) - f(x-h)}{2h} $$
