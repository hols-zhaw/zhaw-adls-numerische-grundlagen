## Pseudozufallszahlen

Die Erzeugung von Zufallszahlen ist ein wichtiger Bestandteil vieler Anwendungen:

- **Simulationen** von biologischen, chemischen oder physikalischen Prozessen, aber auch von Wirtschaftssystemen benötigen Zufallszahlen.
- **Optimierungsverfahren** verwenden Zufallszahlen, um verschiedene Startwerte zu erzeugen und im Suchraum umherzuspringen.
- **Verifikation** von Algorithmen, Software und Hardware wird mithilfe von Testdaten gemacht, die auf Zufallszahlen basieren.
- **Statistische Analysen** benötigen Zufallszahlen, beispielsweise um Signifikanz und Konfidenzintervalle zu schätzen.
- In der **Kryptographie** werden Schlüssel für die Verschlüsselung von Daten mithilfe von Zufallszahlen erzeugt werden.
- In **Computerspielen** werden zufällige Ereignisse und Ergebnisse mit Zufallszahlen generiert.

Echte Zufallszahlen werden durch natürliche Phänomene wie Atmosphärenrauschen, Radioaktivität oder die stochastische Natur der Quantenphysik erzeugt. Diese Zufallszahlen sind wirklich zufällig und können nicht vorhergesagt werden. Sie sind für viele Anwendungen nicht geeignet, da sie nicht praktikabel sind, zu viel Zeit pro Zufallszahl benötigen und nicht reproduzierbar sind. Allerdings können sie in der Kryptographie verwendet werden, um Schlüssel zu erzeugen, die nicht vorhergesagt werden können. Quantencomputer können also wirklich zufällige Zahlen erzeugen, weshalb Quantencomputer im Begriff sind, die Kryptographie zu revolutionieren.

Digitale Computer hingegen können keine wirklich zufälligen Zahlen erzeugen, da sie deterministisch arbeiten. (*Deterministisch* bedeutet, dass für dieselbe Eingabe immer die gleiche Ausgabe erzeugt wird. Das Gegenteil ist *stochastisch*, was bedeutet, dass die Ausgabe nicht vorhersagbar ist.)

[Pseudozufallszahlen](https://de.wikipedia.org/wiki/Pseudozufall) werden durch deterministische Algorithmen generiert, die jedoch zufällig genug für viele Anwendungen sind. Sie sind reproduzierbar und können daher für Simulationen, Tests und statistische Analysen verwendet werden.

Es gibt verschiedene Algorithmen zur Erzeugung von Pseudozufallszahlen, darunter die Middle-Square Methode (MSM), der Linear Congruential Generator (LCG) und der Mersenne Twister.

Bei der Erzeugung von Zufallszahlen ist es wichtig, dass der Algorithmus eine ausreichend lange Sequenz von Zahlen erzeugt, die sich nicht wiederholt, *Periode* genannt. Ausserdem müssen die generierten Zahlen statistisch unabhängig und gleichverteilt sein, um ein zufälliges Verhalten zu gewährleisten.

In der Programmierung können Zufallszahlen durch die Verwendung von speziellen Funktionen oder Bibliotheken erzeugt werden. Diese verwenden unterschiedliche Algorithmen, die unterschiedliche Eigenschaften haben. Die Wahl des Algorithmus hängt von der Anwendung ab.

- Random sampling mit [`numpy.random`](https://numpy.org/doc/stable/reference/random/index.html)
- Zufallsvariablen in ScyPy mit [`scipy.stats`](https://docs.scipy.org/doc/scipy/reference/stats.html#probability-distributions)

## Middle-Square-Methode

Die Middle-Square-Methode (MSM) ist ein einfacher Algorithmus zur Generierung von Pseudozufallszahlen. Er wurde erstmals 1946 von John von Neumann vorgestellt und ist ein Beispiel für einen deterministischen Algorithmus, der auf einem Anfangswert (engl. *seed*) basiert.

Die MSM verwendet eine einfache Formel, um den nächsten Wert in der Zahlenfolge zu berechnen. Zunächst wird der Seed quadriert und das Ergebnis dann in der Mitte abgeschnitten, um eine Zahl mit der gleichen Anzahl an Stellen wie der Seed zu erhalten. Diese Zahl wird dann wiederum verwendet, um auf die gleiche Weise den nächsten Wert in der Folge zu generieren.

Die MSM ist zwar einfach zu implementieren, hat aber einige Nachteile. Je nach Seed ist die Periode (Länge der Sequenz bis sich die Zahlen wiederholen) vergleichsweise kurz. Daher folgen die Zahlen einem einfach verhersehbaren Muster und sind für moderne Anwendugen nicht genügend pseudo-zufällig.

Dennoch ist die MSM ein interessantes Beispiel dafür, wie einfache mathematische Formeln zur Erzeugung von Pseudozufallszahlen verwendet werden können.

#### a) Implementierung Schritt 1

Wir implementieren die Middle-Square-Methode für 4-stellige Zahlen.

Programmiere zuerst eine Funktion namens `msm_next(number)`, die eine ganze Zahl `number` mit 4 Stellen als Parameter nimmt. Sie soll die nächste Zahl in der Zahlenfolge nach dem MSM Algorithmus zurück geben. Validiere deine Implementierung mit den folgenden Beispielen und weiteren Fällen, die du dir selbst ausdenkst:

- $3333 \to 3333^2 = 11108889 \to 1088$
- $1234 \to 1234^2 = 1522756 = 01522756 \to 5227$

*Hinweise:*

- Mit `str()` und `int()` wandelt man Zahlen in Strings und umgekehrt um.
- Ein Teil eines Strings kann mit Slicing extrahiert werden.
- Das Quadrat einer 4-stelligen Zahl hat nicht immer 8 Stellen, wie im Beispiel mit $1234$ zu sehen ist. Führende Nullen können mit `str.zfill()` hinzugefügt werden.

Links:

- [Python String zfill() Method - w3schools](https://www.w3schools.com/python/ref_string_zfill.asp)
- [Python String Methods - Python docs](https://docs.python.org/3/library/stdtypes.html#string-methods)

```python
def msm_next(number):
    """Ein Schritt der Middle-Square-Method für 4-stellige Zahlen."""
    n2_padded = str(number**2).zfill(8)
    return int(n2_padded[2:-2])


(
    msm_next(3333) == 1088,
    msm_next(1234) == 5227,
    msm_next(3839) == 7379,
)
```

#### b) Implementierung Schritt 2

Nun schreibst du eine Funktion namens `msm(seed, n)`, die einen Startwert `seed` und eine Anzahl an Schritten `n` als Parameter nimmt. Sie soll `msm_next()` benutzen, um ausgehend vom Startwert ein NumPy Array von `n` Pseudozufallszahlen nach der MSM zu erzeugen. Das heisst, der Seed kommt nicht im resultierenden Array vor, sondern nur die `n` darauffolgenden Zahlen.

```python
def msm(seed, n):
    """Middle-Square-Method mit Seed und Anzahl Schritten n"""
    number = seed
    sequence = []
    for _ in range(n):
        number = msm_next(number)
        sequence.append(number)
    return np.array(sequence)
```

#### c) Testen der Funktion

### Aufgabe 2: Periode bei Zufallszahlen


Die Periode bei der Erzeugung von Pseudozufallszahlen gibt an, nach wie vielen Schritten sich die Zahlenfolge wiederholt.

Die Periode ist ein Mass für die Qualität eines Zufallszahlengenerators. Eine kurze Periode führt dazu, dass die Zahlen schnell vorhersehbar werden, was schlecht ist. Eine ausreichend lange Periode ist daher wichtig. Im Idealfall ist sie sogar länger als die Anzahl der Zahlen, die der Generator jemals generieren wird.

Es gibt verschiedene Methoden, um die Periode zu erhöhen, z.B. durch Verbesserung des Algorithmus, Kombination mehrerer Methoden oder Erhöhung der Grösse des Zustandsraums (lange Seeds). Die Periode hängt auch von der Wahl des Seeds ab. Daher ist es wichtig, einen geeigneten Seed zu wählen, um sicherzustellen, dass die Periode möglichst gross ist.

#### a) Funktion zur Messung der Periode

Schreibe eine Funktion `msm_period(seed)`, die einen Seed als Parameter erhält. Sie soll die Anzahl Schritte zurückgeben, die der Generator `msm_next()` benötigt, bis er sich zum ersten Mal wiederholt. Nutze dazu eine Menge (`set`), um die bereits generierten Zahlen zu speichern und neue damit zu vergleichen.

```python
def msm_period(seed):
    sequence = set()
    number = msm_next(seed)
    while number not in sequence:
        sequence.add(number)
        number = msm_next(number)
    period = len(sequence) + 1
    return period, number
```

#### b) Verifikation der Periodenmessung

Mit dem Seed 1234 sollte die Periode 57 sein. Die Zahl 0 beginnt sich zu wiederholen. Überprüfe dies mit der Funktion `msm_period()`.

```python
msm_period(1234)
```

#### c) Maximale Periode finden

Berechne die Periode für alle Seeds von 1000 bis 9999 mit einer List Comprehension und finde den Seed mit der längsten Periode mithilfe der Funktion [`max()`](https://docs.python.org/3/library/functions.html#max). Wie lang ist diese längste Periode?

```python
periods = [(msm_period(seed)[0], seed) for seed in range(1000, 10000)]
max_period, max_seed = max(periods)

print(f"Die maximale Periode ist {max_period} und wird erreicht mit Seed {max_seed}.")
```

Alternativ kann man auch `np.argmax()` verwenden, um den Index der längsten Periode zu finden:

```python
seeds = np.arange(1000, 10000)
periods, _ = np.vectorize(msm_period)(seeds)
i_max = np.argmax(periods)

print(
    f"Die maximale Periode ist {periods[i_max]} und wird erreicht mit Seed {seeds[i_max]}."
)
```
