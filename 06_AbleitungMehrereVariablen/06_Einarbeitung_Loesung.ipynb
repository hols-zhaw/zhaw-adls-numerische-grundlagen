### Partielle Ableitung

Ableiten nach mehreren Variablen beschreibt den Prozess, bei dem man die Änderung einer Funktion in Bezug auf die Änderungen mehrerer unabhängiger Variablen untersucht. Dies wird häufig beim Modellieren von dynamischen Systemen und in der Optimierung verwendet. Um dies zu tun, verwendet man die sogenannten **partiellen Ableitungen**, welche die Änderung einer Funktion in Bezug auf eine einzelne unabhängige Variable beschreiben.

### Arbeitsauftrag

#### 1. Theorie

Schreibe für dich eine Zusammenfassung der vorgehensweise bei einer partiellen Ableitung auf.

**Lösung:**

- Bei einer partiellen Ableitung leitet man nur eine Variable einer Funktion mit mehreren Variablen ab. 

- Bei der partiellen Ableitung wird nach einer der unabhängigen Variablen abgeleitet (zum Beispiel $x$ oder $y$). Die anderen Variablen werden in diesem Prozess wie Konstanten behandelt.

#### 2. Ableitungsregeln

Bei der partiellen Ableitung gelten die gleichen Regeln wie bei der Ableitung in einer Variablen.

Fasse die Ableitungsregeln für Funktionen von einer Variablen kurz für dich zusammen. (Siehe Unterlagen aus der Analysis.)

**Lösung:**

- Summen- und Faktorregel:
    $$ f(x) = a\cdot g(x) + b\cdot h(x) \qquad\Rightarrow\qquad f'(x) = a\cdot g'(x) + b\cdot h'(x) $$

- Produktregel:
    $$ f(x) = g(x) \cdot h(x) \qquad\Rightarrow\qquad f'(x) = g'(x) \cdot h(x) + g(x) \cdot h'(x) $$

- Quotientenregel:
    $$ f(x) = \dfrac{g(x)}{h(x)} \qquad\Rightarrow\qquad f'(x) = \dfrac{g'(x) \cdot h(x) - g(x) \cdot h'(x)}{(h(x))^2} $$

- Kettenregel:
    $$ f(x) = v(w(x)) \qquad\Rightarrow\qquad f'(x) = v'(w(x)) \cdot w'(x) $$

#### 3. Elementare Ableitungen

Die Ableitungen der folgenden Funktionen sind wichtig und sollten bekannt sein:

- Potenzfunktionen: $\displaystyle \Big[ x^p \Big]' = p \cdot x^{p-1} \\[1ex]$
- Exponentialfunktionen: $\displaystyle \Big[ b^x \Big]' = \ln(b) \cdot b^x \\[1ex]$
- natürliche Exponentialfunktion: $\displaystyle \Big[ e^x \Big]' = e^x \\[1ex]$
- Logarithmusfunktionen: $\displaystyle \Big[ \log_b(x) \Big]' = \frac{1}{\ln(b) \cdot x} \\[1ex]$
- natürlicher Logarithmus: $\displaystyle \Big[ \ln(x) \Big]' = \frac{1}{x} \\[1ex]$
- Sinus: $\displaystyle \Big[ \sin(x) \Big]' = \cos(x) \\[1ex]$
- Cosinus: $\displaystyle \Big[ \cos(x) \Big]' = -\sin(x)$

#### 4. Erste Beispiele

Im Folgenden sind einige Beispiele für partielle Ableitungen gegeben. Bestimme welche der Ableitungsregeln benutzt wurden.

**a)** $\displaystyle f(x_1,x_2) = 3x_1^3 + 4x_1x_2^2 + 4x_2$

\begin{align*}
    \dfrac{\partial f}{\partial x_1}(x_1,x_2) &= 9x_1^2 + 4x_2^2 \\[2ex]
    \dfrac{\partial f}{\partial x_2}(x_1,x_2) &= 8x_1x_2 + 4
\end{align*}

**a) Lösung:** Summenregel und Faktorregel

**b)** $\displaystyle g(x,y) = 5x^3 \cdot \sin(x) \cdot 4y^2 \cdot \cos(y)$

\begin{align*}
    \frac{\partial g}{\partial x}(x,y) &= \left( 15x^2\cdot \sin(x) + 5x^3\cdot \cos(x) \right) \cdot 4y^2\cdot \cos(y) \\[2ex]
    \frac{\partial g}{\partial y}(x,y) &= 5x^3\cdot \sin(x) \cdot \left( 8y \cdot \cos(y)- 4y^2\cdot \sin(y) \right)
\end{align*}

**b) Lösung:** Faktorregel und Produktregel

**c)** $\displaystyle h(x,y) = \dfrac{x^2 + 3y}{5xy}$

\begin{align*}
    \frac{\partial h}{\partial x}(x,y) &= \frac{2x \cdot 5xy - \left(x^2+3y\right) \cdot 5y}{\left(5xy\right)^2} \\[2ex]
    \frac{\partial h}{\partial y}(x,y) &= \frac{3 \cdot 5xy - \left(x^2+3y\right) \cdot 5x}{\left(5xy\right)^2}
\end{align*}

**c) Lösung:** Quotientenregel, Summenregel und Faktorregel

**d)** $\displaystyle N(t,x) = \exp\left(-t \cdot x^2\right)$

\begin{align*}
    \frac{\partial N}{\partial t}(t,x) &= \exp\left(-t \cdot x^2\right) \cdot \left(-x^2\right) \\[2ex]
    \frac{\partial N}{\partial x}(t,x) &= \exp\left(-t \cdot x^2\right) \cdot \left(-2tx\right)
\end{align*}

**d) Lösung:** Kettenregel und Faktorregel

#### 5. Notation

Partielle Ableitungen schreibt man normalerweise mit dem Zeichen $\partial$ statt $d$. Oder man verwendet eine Kurznotation, bei der man die Variable, nach der
abgeleitet wird, tiefgestellt neben den Funktionsnamen schreibt (Subskript). Somit gibt es drei Schreibweisen:

$$
    \frac{\partial f}{\partial x} = \frac{\partial }{\partial x}f = f_x
$$

Bemerke: Keine dieser Schreibweisen beinhalten einen Strich, wie bei der Ableitung in einer Variablen, $f'(x)$.

Bei zweite und höheren Ableitungen wird $\partial$ bzw. $\partial x$ entsprechend potenziert. Beispiel:

$$
    \frac{\partial }{\partial x}\left(\frac{\partial }{\partial x}f\right) = \frac{\partial^2 f}{\partial x^2} = \frac{\partial }{\partial x}f_x = f_{xx}
$$

und

$$
    \frac{\partial }{\partial x}\left(\frac{\partial }{\partial y}f\right) = \frac{\partial^2 f}{\partial x \partial y} = \frac{\partial }{\partial y}f_x = f_{xy}
$$

**c)** Berechne $\frac{\partial}{\partial x}\left( \frac{\partial}{\partial y}\left[ \frac{ay}{x} \right] \right)$ und umgekehrt $\frac{\partial}{\partial y}\left( \frac{\partial}{\partial x}\left[ \frac{ay}{x} \right] \right)$. Vergleiche die beiden Resultate. Was fällt auf?

**c) Lösung:**

\begin{align*}
    \frac{\partial}{\partial x}\left( \frac{\partial}{\partial y}\left[ \frac{ay}{x} \right] \right)

    &= \frac{\partial}{\partial y}\left( -\frac{ay}{x^2} \right) \\
    &= -\frac{a}{x^2}
\end{align*}

Die Reihenfolge der Ableitungen spielt bei höheren Ableitungen keine Rolle.

#### 7. Interpretation

In den folgenden Aufgaben geht es um das Verständnis des Ergebnisses einer partiellen Ableitung.

**a)** Körpergewicht

Angenommen dein Gewicht $m$ in kg ist eine Funktion $m(c,t)$ der täglich eingenommenen Kilokalorien $c$ und der Minuten $t$, die du täglich trainierst.

Interpretiere die folgenden Angaben, indem du Einheiten für $m$, $c$ und $t$ verwendest:

$$
\frac{\partial m}{\partial c}(2500,30) = 0.02 \,,\qquad \frac{\partial m}{\partial t}(2500,30) = -0.025
$$

**a) Lösung:**

- $\dfrac{\partial m}{\partial c}$ hat die Einheit kg/kcal. Also würde man, wenn man täglich 30 min trainiert und 2500 kcal zu sich nimmt, pro zusätzliche Kilokalorie 0.02 kg zunehmen.

- $\dfrac{\partial m}{\partial t}$ hat die Einheit kg/min. Also würde man, wenn man täglich 30 min trainiert und 2500 kcal zu sich nimmt, pro zusätzlicher Trainingsminute 0.025 kg abnehmen.

**b)** Ansteckungsrisiko

Nehmen wir an, die Virenbelastung der Luft in der Nähe einer erkrankten Person nehme wie

$$
    v(d) = \exp\left(-\tfrac{1}{5}d^2\right)
$$

mit der Distanz $d$ von der infektiösen Person ab. (Das ist ein einfaches Modell, das nicht den Anspruch hat, die Realität korrekt abzubilden.)

Zudem nehme das Ansteckungsrisiko $r$ linear mit der Virenbelastung $v$ und der ihr ausgesetzen Zeit $t$ zu, also

$$
    r(t, d) = t \cdot v(d) = t \cdot \exp\left(-\tfrac{1}{5}d^2\right)
$$

(Bemerkung: Diese Funktion kann nicht direkt als Wahrscheinlichkeitsdichte interpretiert werden, da sie nicht korrekt normiert ist.)

Ein Bild sagt wohl mehr als die Funktionsgleichung:

```python
import numpy as np
import matplotlib.pyplot as plt


def r(t, d):
    return t * np.exp(-(d**2) / 5)


t = np.linspace(0, 10, 100)
d = np.linspace(0, 5, 100)
T, D = np.meshgrid(t, d)
R = r(T, D)
plt.contourf(T, D, R, 20)
plt.plot(8, 2, "w.")
plt.text(8, 2.2, "(8, 2)", color="white", weight="bold")
plt.xlabel("Aufenthaltsdauer t [s]")
plt.ylabel("Distanz d [m]")
plt.xlim(t.min(), t.max())
plt.ylim(d.min(), d.max())
plt.colorbar(label="Ansteckungsrisiko r [willkürliche Einheit]")
plt.title("Ansteckungsrisiko in Abhängigkeit von Zeit und Distanz")
plt.show()
```

Hier nochmals die Funktion: $r(t,d) = t \cdot \exp\left(-d^2/5\right)$. Das Risiko sei eine einheitslose Grösse.

1. Berechne $\dfrac{\partial r}{\partial t}$ und $\dfrac{\partial r}{\partial d}$.

2. Werte die Funktion $r$ und die partiellen Ableitungen an der Stelle $(t, d) = (8, 2)$ aus. (Dieser Punkt ist im Diagramm eingezeichnet.)

3. Bestimme die Einheiten der partiellen Ableitungen. (Notation: $[x]$ bedeutet "die Einheit der Variable $x$".)

4. Beschreibe die Bedeutung dieser Grössen.

**Lösungen:**

**1.**
$$
\begin{align*}
    \dfrac{\partial r}{\partial t} &= \exp\left(-\tfrac{1}{5}d^2\right) \\[2ex]
    \dfrac{\partial r}{\partial d} &= t\cdot \exp\left(-\tfrac{1}{5}d^2\right) \cdot \left(-\tfrac{2}{5}d\right)
\end{align*}
$$

**2.**
$$
\begin{align*}
    r(8,2) &= 8\cdot \exp\left(-\tfrac{1}{5}2^2\right) \approx 3.59 \\[2ex]
    \frac{\partial r}{\partial t}(8,2) &\approx 0.45 \\[2ex]
    \frac{\partial r}{\partial d}(8,2) &\approx -2.9
\end{align*}
$$

**3.**
$$
\begin{align*}
    \left[\dfrac{\partial r}{\partial t}\right] &= \frac{[r]}{[t]} = \text{s}^{-1} \\[2ex]
    \left[\dfrac{\partial r}{\partial d}\right] &= \frac{[r]}{[d]} = \text{m}^{-1}
\end{align*}
$$

**4.**
- $\dfrac{\partial r}{\partial t}(8,2)\approx 0.45\,\text{s}^{-1}$ bedeutet, dass sich das Ansteckungsrisiko bei $(t,d)=(8,2)$ um 0.45 (oder 12.5%) pro Sekunde erhöht, wenn man stehen bleibt.

- $\dfrac{\partial r}{\partial d}(8,2)\approx -2.9\,\text{m}^{-1}$ bedeutet, dass sich das Ansteckungsrisiko bei $(t,d)=(8,2)$ um 2.9 (oder 80.8%) oder pro Meter verringert, wenn man sich instantan weiter weg bewegen würde.
