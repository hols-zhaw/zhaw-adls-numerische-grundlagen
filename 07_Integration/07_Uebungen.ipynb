```python
import numpy as np
import matplotlib.pyplot as plt


def trapz(func, a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = func(x)
    return (h / 2) * (y[0] + 2 * np.sum(y[1:-1]) + y[-1])


def trapz_data(x, y):
    return np.sum((x[1:] - x[:-1]) * (y[1:] + y[:-1])) / 2
```
