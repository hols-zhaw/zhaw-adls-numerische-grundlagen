### Funktionen mit mehreren Variablen


Viele Grössen in naturwissenschaftlichen und technischen Kontexten hängen von mehreren Variablen ab:

-   Die Temperatur in einem Raum hängt von der 3D-Position $(x,y,z)$ ab, oder auch noch von der Zeit $(x,y,z,t)$.
-   Das Wachstum einer Zellkutur hängt nicht nur von der Zeit $t$ ab, sondern auch von der Menge der vorhandenen Nährstoffe, vom pH-Wert der Umgebung und wohl noch von anderen Variablen.

So gibt es auch in der Mathematik Funktionen, die von mehreren Variablen abhängen. Häufig heissen die unabhängigen Variablen $x$, $y$ und $z$, können aber auch mit einem Index durchgezählt ($x_1$, $x_2$, ...) oder in einem Vektor $\vec x \in \mathbb{R}^n$ zusammengepackt werden. Ein Beispiel:

$$
    f(x,y) = x^2 - y^2
$$

Das heisst, jedem Paar $(x,y)$ wird der Wert $x^2-y^2$ zugeordnet, z.B.

$$
    f(4,5) = 4^2 - 5^2 = 16 - 25 = -9
$$

Das Wertepaar $(x,y)$ kann dabei als Vektor oder Punkt in der Ebene verstanden werden: $\vec x = (x,y)$. Der Funktionswert kann als Höhe einer Landschaft verstanden werden, die über der Ebene liegt.

### Partielle Funktionen (Isolinien)


Wertet man eine Funktion von $(x,y)$ für einen spezifischen, konstanten Wert für $x$ aus und lässt $y$ weiterhin variabel, so führt dies zu einer _partiellen Funktion_ von $y$:

$$ p_3(y) = f(3, y) = 3^2 - y^2 = 9 - y^2 $$

$$ p_5(y) = f(5, y) = 5^2 - y^2 = 25 - y^2 $$

Funktion für konstantes $y$ auswerten führt zu einer _partiellen Funktion_ von $x$:

$$ q_2(x) = f(x, 2) = x^2 - 2^2 = x^2 - 4 $$

$$ q_4(x) = f(x, 4) = x^2 - 4^2 = x^2 - 16 $$

Die partiellen Funktionen sind Schnitte entlang konstanter Werte für alle bis auf eine der unabhängigen Variablen. Diese werden auch _Isolinien_ genannt.


```python
import numpy as np
import matplotlib.pyplot as plt


def f(x, y):
    return x**2 - y**2


y = np.linspace(-2, 2)
p3 = f(3, y)
p5 = f(5, y)

x = np.linspace(-2, 2)
q2 = f(x, 2)
q4 = f(x, 4)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4), sharey=True)

ax1.plot(y, p3, label="$x=3$")
ax1.plot(y, p5, label="$x=5$")
ax1.set_xlabel("$y$")
ax1.set_ylabel("$p_x$")
ax1.legend()
ax1.grid()
ax1.set_title("Isolinien für konstante $x$")

ax2.plot(x, q2, label="$y=2$")
ax2.plot(x, q4, label="$y=4$")
ax2.set_xlabel("$x$")
ax2.set_ylabel("$q_y$")
ax2.legend()
ax2.grid()
ax2.set_title("Isolinien für konstante $y$")

plt.tight_layout()
plt.show()
```

### Höhenlinien

#### Grundsätzliches Vorgehen


Es sollen Linien konstanter Funktionswerte $z$ in der $x$-$y$-Ebene dargestellt werden.

1. Die Funktionsgleichung $z=f(x,y)$ entweder nach $x$ oder nach $y$ auflösen, je nach dem was einfacher ist. Falls es mehrere Lösungen gibt, z.B. beim Wurzelziehen, dann müssen alle diese Lösungen dargestellt werden.
2. Bereich in $x$ (bzw. $y$) bestimmen, in dem die Funktion dargestellt werden soll. Dieser hängt stark vom Kontext ab. Array mit $x$-Werten (bzw. $y$-Werten) definieren.
3. Zugehörige $y$-Werte (bzw. $x$-Werte) für eine Serie von $z$-Werten berechnen und darstellen.

### Aufgabe 2: Darstellung mit Matplotlib


**2. a)** Stelle die Höhenlinie zu $z=1$ im Bereich $x\in[0,3]$ mit Matplotlib dar.


```python
def y_z(z, x):
    return z - 2 * x**2


x = np.linspace(0, 3)
y = y_z(1, x)

plt.plot(x, y, label="z = 1")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.title("Höhenlinie (konstantes z)")
plt.show()
```

**2. b)** Nun machen wir das gleiche für mehrere $z$-Werte, sagen wir $z\in\{-5, -3, 0, 3, 5\}$ (`levels`), und stellen diese zusammen in einem Diagramm dar.

Füge eine Legende mit den zugehörigen Werten von $z$ hinzu.


```python
# Aufgabe 2b:

levels = (-5, -3, 0, 3, 5)
x = np.linspace(0, 3)

for z in levels:
    y = y_z(z, x)
    plt.plot(x, y, label=f"z = {z}")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.title("Höhenlinien (konstante z)")
plt.show()
```

### Aufgabe 3: Erstes Beispiel mit `plt.contour`


Meist kennt man anstatt der Funktionswerte eher den Bereich in der $x$-$y$-Ebene, für den man sich interessiert. Oder man kann die Funktion nicht von Hand nach einer der Variablen auflösen. Oder man hat Daten statt eine analytische Funktion gegeben. Zudem könnte es schwierig sein, die $z$-Werte (Levels) selbst zu wählen. In all diesen Fällen hilft einem die Funktion [`plt.contour`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.contour.html).

Schau dir das erste Beispiel in der [Contour Demo](https://matplotlib.org/stable/gallery/images_contours_and_fields/contour_demo.html) an.

Stelle nun die Höhenlinien der Funktion $f(x,y)$ aus Aufgabe 1 im Bereich $x\in[0,3]$ und $y\in[-6,6]$ mithilfe von `plt.contour` dar inkl. Beschriftung der Achsen und der Höhenlinien.


```python
x = np.linspace(0, 3)
y = np.linspace(-6, 6)
X, Y = np.meshgrid(x, y)
Z = 2 * X**2 + Y

cs = plt.contour(X, Y, Z)
plt.clabel(cs)
plt.colorbar(label="z")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.title("Höhenlinien für $f(x, y) = 2x^2 + y$")
plt.show()
```

Warum unterscheidet sich diese Darstellung von der in Aufgabe 2?


1. Die dargestellten $z$-Werte (Levels) sind nicht dieselben. Wir könnten diese mit der Option `levels` in `plt.contour` beinflussen.
2. Der dargestellte Bereich in $y$ ist nicht der gleiche.
