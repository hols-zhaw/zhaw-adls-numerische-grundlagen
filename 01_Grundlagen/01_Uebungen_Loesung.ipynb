#### 1.1 Temperatur in Davos

**a)** Daten importieren

Die Datei `davos_temperatur.txt` beinhaltet drei Spalten. Die erste gibt das Kürzel einer Wetterstation an, die zweite einen Zeitstempel (Datum und Stunde: `YYYYMMDDHH`) und die dritte die gemessene Temperatur.

Importiere die dritte Spalte in ein Array namens `temperature`. Gib den Shape des Arrays und die ersten 5 Elemente aus.

```python
temperature = np.loadtxt("davos_temperatur.txt", delimiter=";", skiprows=1, usecols=2)
print("shape:", temperature.shape)
print(temperature[:5])
```

**b)** Daten darstellen

- Stelle den Temperaturverlauf mit einem Liniendiagramm grafisch dar.
  - Als Zeitachse (horizontal) soll die Anzahl Stunden nach Messbeginn benutzt werden. Diese entspricht einfach dem Index des Temperaturarrays. 
- Schränke den Bereich der Zeitachse auf die Stunden 6 - 24 ein. Benutze dazu die Funktion [`plt.xlim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.xlim.html).
- Beschrifte die Achsen und gib dem Diagramm einen Titel.

```python
# Messzeitpunkte erstellen
time = np.arange(0, len(temperature))

# Temperaturen plotten
plt.plot(time, temperature)

# x-Achse einschränken
plt.xlim(6, 24)

# Achsenbeschriftungen setzen
plt.xlabel("Messzeitpunkt [h]")
plt.ylabel("Temperatur [°C]")

# Titel setzen
plt.title("Temperatur in Davon zwischen 6 und 24 Uhr")

# Diagramm anzeigen
plt.show()
```

#### 1.3 Bakterien

**a)** Daten importieren

- Importiere die Daten in der Datei `bacteria_nutrient.csv` in ein Array namens `bac_data`.
- Gib den Shape des Arrays und die ersten 5 Zeilen aus.
- Wie viele Beobachtungen sind es und welche Variablen sind in den Spalten zu finden?
- Übertrage die Spalten in entsprechend benannte 1-D Arrays.

```python
# Daten aus der CSV-Datei laden
bac_data = np.loadtxt("bacteria_nutrient.csv", delimiter=",", skiprows=1)
print("shape:", bac_data.shape)
print(bac_data[:5, :])
```

**b)** Bakterienanzahl und Nährstoffmenge zusammen darstellen

- Stelle den Verlauf der Bakterienanzahl und der Nährstoffmenge über die Zeit dar. Erzeuge dazu mit [`plt.subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html) zwei übereinanderliegende Koordinatensysteme. Stelle die Bakterienanzahl im oberen und die Nährstoffmenge im unteren dar.
- Beschrifte die Achsen und gib dem gesamten Diagramm einen Titel (diesmal mit [`plt.suptitle`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.suptitle.html)).
- Schränke den Darstellungsbereich auf Min/Max der Zeitwerte ein. Benutze dazu [`ax.set_xlim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xlim.html).
- Füge in beiden Koordinatensystemen ein Gitternetz hinzu mit [`ax.grid`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.grid.html).

```python
fig, (ax0, ax1) = plt.subplots(2, 1, sharex=True)
ax0.plot(time, bacteria)
ax0.set_ylabel("Bakterienanzahl")
ax0.grid()
ax1.plot(time, nutrient)
ax1.set_ylabel("Nährstoffmenge")
ax1.set_xlabel("Zeit")
ax1.grid()
ax1.set_xlim(time.min(), time.max())
plt.suptitle("Bakterienwachstum und Nährstoffkonzentration über die Zeit")
plt.show()
```

**c)** Hinweis

Zwei Graphen unterschiedlicher Einheiten (wie Bakterienanzahl und Nährstoffmenge) sollten wenn möglich nicht im gleichen Diagramm (Koordinatensystem) dargestellt werden. Die Variante mit den zwei Subplots wie oben ist vorzuziehen. Hier wäre zur Vollständigkeit noch eine Variante mit zwei y-Achsen (links und rechts).

```python
fig, ax1 = plt.subplots()

# Plot der Bakterienanzahl
color = "tab:red"
ax1.plot(time, bacteria, color=color)
ax1.tick_params(axis="y", labelcolor=color)
ax1.set_ylabel("Bakterienanzahl", color=color)
ax1.set_xlabel("Zeit")
ax1.set_xlim(time.min(), time.max())

# Zweites Koordinatensystem für Nährstoffmenge darüberlegen
color = "tab:blue"
ax2 = ax1.twinx()
ax2.plot(time, nutrient, color=color)
ax2.tick_params(axis="y", labelcolor=color)
ax2.set_ylabel("Nährstoffmenge", color=color)

plt.suptitle("Bakterienwachstum und Nährstoffkonzentration über die Zeit")
plt.tight_layout()
plt.show()
```

#### 2.3 Eigenwerte und Eigenvektoren

**a)** Berechne die Eigenwerte und Eigenvektoren der Matrix

$$
\begin{pmatrix}
1 & 2 \\
1 & 0
\end{pmatrix}
$$

Benuzte dazu die Funktion [`np.linalg.eig`](https://numpy.org/doc/stable/reference/generated/numpy.linalg.eig.html). Welche Eigenwerte und Eigenvektoren gehören jeweils zusammen?

```python
evals, evecs = np.linalg.eig([[1, 2], [1, 0]])

for i, l in enumerate(evals):
    print(f"Eigenwert {i+1}: {l:.3},\tEigenvektor {i+1}: {evecs[:,i]}")
```

**b)** berechne die Eigenwerte und Eigenvektoren der $4 \times 4$ Matrix aus Aufgabe 5a).

### Aufgabe 3: Daten simulieren (Zufallszahlen)

#### 3.1 Zufallszahlen erzeugen

- Erstelle einen Zufallszahlengenerator mit `np.random.default_rng()` mit einem selbstgewählten Seed.
- Erzeuge mit [`numpy.random.Generator.uniform`](https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.uniform.html) 250 **gleichverteilte** Zufallszahlen zwischen -2 und 2. Speichere sie im Array `x`. 
- Erzeuge mit [`numpy.random.Generator.normal`](https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.normal.html) 250 **normalverteilte** Zufallszahlen mit Mittelwert 5 und Standardabweichung 2. Speichere sie im Array `y`.
- Stelle (x,y) als Punktwolke dar. Nutze dazu die Funktion [`plt.scatter`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html).

```python
rng = np.random.default_rng(987234)
x = rng.uniform(-2, 2, 250)
y = rng.normal(5, 2, 250)

plt.scatter(x, y, marker=".")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

#### 3.2 Daten mit Messfehlern (Rauschen, Noise) simulieren

Wir betrachten die folgende Sinusfunktion als Signal

$$ y = 2 \sin(2 \pi t) + 10 $$

und addieren ein zufälliges Rauschen dazu.

**a)** Signal erzeugen

Definiere für die Zeit `t` mit [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) ein Array von $0$ bis $5$. Dann berechne die Funktion $y$ in diesem Bereich.

```python
t = np.linspace(0, 5)
y = 2 * np.sin(2 * np.pi * t) + 10
```

**b)** Rauschen hinzufügen

- Erzeuge normalverteilte Messfehler (Zufallszahlen) mit Mittelwert $0$ und Standardabweichung $0.8$ in einem Array namens `noise`.
- Addiere die Messfehler zu `y` und speichere die simulierten Daten in `noisy`.

```python
rng = np.random.default_rng(12345678)
noise = rng.normal(0, 0.8, size=y.shape)
noisy = y + noise
```

**c)** Signal mit und ohne Rauschen darstellen

- Stelle `y` und `noisy` über die Zeit im gleichen Diagramm dar.
- Die Theorie `y` soll mit einer verbundenen Linie dargestellt werden, die Messwerte `noisy` nur als Punkte.
- Füge mit [`plt.legend`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.htmlml) eine Legende hinzu.
- Schreibe Achsen an.

```python
plt.plot(t, y, label="Theorie")
plt.plot(t, noisy, "--o", label="Messwerte")
plt.xlabel("t")
plt.ylabel("y")
plt.xlim(t.min(), t.max())
plt.legend()
plt.grid()
plt.title("Signal mit und ohne Rauschen")
plt.show()
```

**d)** Daten exportieren

Speichere die drei Variablen `t`, `y` und `noisy` in eine CSV-Datei `noisy.csv` ab.

- Für das Zusammenfügen mehrerer Spalten in ein 2-D Array gibt es die Funktion [`np.column_stack`](https://numpy.org/doc/stable/reference/generated/numpy.column_stack.html).
- Benutze [`np.savetxt`](https://numpy.org/doc/stable/reference/generated/numpy.savetxt.html) für den Datenexport.
  - Benutze Strichpunkte als Trennzeichen, damit die Datei auch problemlos in einem deutschsprachigen Excel geöffnet werden kann.
  - Schaue in der Dokumentation nach, wie man in der ersten Zeile der CSV-Datei Spaltenüberschriften hinzufügen kann.

```python
data = np.column_stack((t, y, noisy))
np.savetxt("noisy.csv", data, delimiter=";", header="Zeit;Theorie;Messwert")
```
