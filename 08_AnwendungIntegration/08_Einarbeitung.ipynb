## Arithmetisches Mittel einer Funktion

### Input

Für eine Messreihe $\vec y = (y_0,\ldots,y_{n-1})$ mit $n$ Messwerten $y_i$ gleichmässig über die Zeit verteilt (also $t_i = t_0 + i h$) ist das arithmetische Mittel

$$ \overline{y} = \frac{1}{n}\sum_{i=0}^{n-1} y_i $$

eine Schätzung für den Mittelwert der Funktion $y=f(t)$, die der Messreihe zugrunde liegt.

Analog dazu ist das arithmetische Mittel dieser Funktion $f:D\to\mathbb{R}$ in einem abgeschlossenen Intervall $[a,b]\subseteq D$ definiert als

$$ \overline{f} = \frac{1}{b-a}\int_a^b f(t) \,\mathrm{d}t $$

Vergleiche dazu Aufgabe 4.d) aus dem Übungsblatt 07 Integration.

## Gewichtetes arithmetisches Mittel einer Funktion

### Input

Für eine Messreihe $\vec{y} = (y_0,\ldots,y_{n-1})$ mit $n$ Messwerten $y_i$ ist das **gewichtete arithmetische Mittel**

$$ \overline{y} = \sum_{i=0}^{n-1} w_i y_i $$

wobei die Gewichte $w_i$ normalerweise normiert werden, so dass $\sum_{i=0}^{n-1} w_i = 1$. Sind sie dies noch nicht, so normiert man die Gewichte, indem man sie durch die Summe aller Gewichte teilt:

$$ w_i \ \to \ \frac{w_i}{\sum_{i=0}^{n-1} w_i} $$

Beispiel: Zwei Prüfungen während des Semesters zählen je 20% der Modulnote und die abgesetzte Prüfung zählt 60%. Die Gewichtung der Prüfungsnoten $y_i$ sind somit $w_0 = 20$, $w_1 = 20$, $w_2 = 60$. Die Normierung der Gewichte ergibt $\sum_{i=0}^{n-1} w_i = 100$. Die normierten Gewichte sind also $w_0 = 0.2$, $w_1 = 0.2$, $w_2 = 0.6$.

Beobachtung: Das gleichgewichtete (normale) artihmetische Mittel erhält man mit $w_i = \frac{1}{n}$ für $i=0,\ldots,{n-1}$.



Für die kontinuierliche Funktion $f(x)$ kann analog dazu das gewichtete arithmetische Mittel im Intervall $[a,b]$ definiert werden:

$$ \overline{f} = \int_a^b w(t) f(t) \,\mathrm{d}t $$

Die Gewichtungsfunktion (*engl.* weight function) $w(t)$ ist auch hier meist normiert, so dass

$$ \int_a^b w(t) \,\mathrm{d}t = 1 $$

gilt. Ist die Gewichtung noch nicht normiert, kann sie wie folgt normiert werden:

$$ w(t) \ \to \ \frac{w(t)}{\int_a^b w(t) \,\mathrm{d}t} $$

Die Gewichtungsfunktion kann je nach Anwendung etwas anders interpretiert werden, siehe Vertiefung.

## Gleitender Mittelwert

### Input

Wir beginnen wieder mit einer Messreihe $\vec y = (y_0,\ldots,y_{n-1})$ mit $n$ Messwerten $y_i$ gleichmässig über die Zeit verteilt ($t_i=t_0+ih$). Um die Messwerte zu glätten, berechnet man häufig den [gleitenden Mittelwert](https://de.wikipedia.org/wiki/Gleitender_Mittelwert) über die letzten $m$ Messwerte (*engl.* simple moving average, SMA). Aus Sicht eines Zeitpunkts $t_i$ mit $i\geq m$ heisst das, man mittelt über diesen und die $m-1$ vorhergehenden Messwerte:

$$ \mathrm{SMA}^{(m)}_i = \frac{1}{m}\left(y_i + y_{i-1} + \ldots + y_{i-(m-1)}\right) = \frac{1}{m} \sum_{k=0}^{m-1} y_{i-k} \qquad\text{für}\quad m-1 \leq i < n $$

Die folgende Darstellung zeigt die Messreihe $\vec y$ und die Folge der gleitenden Mittelwerte $\mathrm{SMA}^{(3)}$. (Vorsicht, hier wurden die Werte $\mathrm{SMA}^{(3)}_i$ jeweils auf die Mitte des gemittelten Tripels zentriert, was nicht ganz der obigen Definition entspricht).


> ![Gleitender Mittelwert](gleitender_mittelwert_ungewichtet.png)
>
> [Gut informiert](https://de.wikipedia.org/wiki/Benutzer:Gut_informiert) (2012, Dezember 4). *ittelwertSignalvergleich3* [SVG]. Wikimedia Commons (https://commons.wikimedia.org/wiki/File:MittelwertSignalvergleich3.svg). Gemeinfrei.


Nun können wir auch hier die zu mittelnden Messwerte mit Gewichten $w_k$ versehen, so dass sich der gewichtete gleitende Mittelwert (*engl.* weighted moving average, WMA) ergibt:

$$
    \mathrm{WMA}^{(m)}_i = \sum_{k=0}^{m-1} w_k y_{i-k}
$$

Im folgenden Beispiel mit $m=3$ sind die Gewichte "Dreiecksförmig" (Binomialfilter) gewählt:

$$
    \vec w = \frac{1}{4} \left( 1,\, 2,\, 1 \right) \qquad \Rightarrow \qquad \mathrm{WMA}^{(3)}_i = \frac{1}{4}\left(y_i + 2y_{i-1} + y_{i-2}\right)
$$

Beobachtung: Mit dieser Wahl der Gewichte ist die Überkompensation bei starken Schwankungen (hier bei $8 < t < 12$) wesentlich weniger ausgeprägt.
