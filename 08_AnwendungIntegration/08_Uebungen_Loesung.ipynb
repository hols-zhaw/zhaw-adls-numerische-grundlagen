### Aufgabe 2: Gewichteter Mittelwert

#### 2.1) Gewichtungsfunktion normieren

Normiere die folgenden Gewichtungsfunktion über das angegebene Intervall.

1. $w(x) = x$ im Intervall $[1,3]$

2. $w(t) = -t(t-1)$ im Intervall $[0,1]$

3. $w(t) = e^{-t}$ im Intervall $[0,\infty[$. (*Hinweis:* Berechne das Integral im Intervall $[0,x]$ und nehme dann den Grenzwert für $x\to\infty$.)

**Lösungen**:

1.  

$$
\int_{1}^{3} x \,\mathrm{d}x
= \frac{1}{2} \bigg[ x^2 \bigg]_{1}^{3}
= \frac{1}{2} \left( 9 - 1 \right)
= 4
$$

$$ \quad\Rightarrow\quad w(x) = \frac{1}{4}x $$

2. 

$$
\int_{0}^{1} -t(t-1) \,\mathrm{d}t
= \int_{0}^{1} t - t^2  \,\mathrm{d}t
= \bigg[ \frac{1}{2}t^2 - \frac{1}{3}t^3 \bigg]_{0}^{1}
= \frac{1}{2} - \frac{1}{3}
= \frac{1}{6}
$$

$$ \quad\Rightarrow\quad w(t) = 6 \left(t - t^2\right) $$

3. 

$$
\int_{0}^{\infty} e^{-t} \,\mathrm{d}t
= \lim_{x\to\infty} \bigg[ -e^{-t} \bigg]_{0}^{x}
= \lim_{x\to\infty} 1 - e^{-x}
= 1
$$

Die Funktion $w(t)=e^{-t}$ ist bereits normiert.

&= -\frac{1}{2} e^{-x} \left(\sin(x) + \cos(x)\right) + \frac{1}{2}e^{-0} \left(\sin(0) + \cos(0)\right) \\
&= -\frac{1}{2} e^{-x} \left(\sin(x) + \cos(x)\right) + \frac{1}{2}

$$
\int_{0}^{\infty} w(t) h(t) \,\mathrm{d}t

### Aufgabe 3: Skalarprodukt zweier Funktionen

#### 3.1) Eigenschaften des Skalarprodukts

1.  Zeige, dass das Skalarprodukt zweier Funktionen

$$ \langle f, g \rangle = \int_a^b f(x) g(x) \,\mathrm{d}x $$

unter der Annahme, dass das Integral konvergiert (berechenbar ist), die folgenden Eigenschaften erfüllt:

1. Kommutativität: $\langle f, g \rangle = \langle g, f \rangle $
   
2. Linearität ($a$ konstant): $\langle f + a \cdot g, h \rangle = \langle f, h \rangle + a \langle g, h \rangle $

**Lösungen:**

1. Folgt aus der Kommutativität (Vertauschbarkeit) des Produkts von reellen Zahlen: $x \cdot y = y \cdot x$ für alle $x,y\in\mathbb{R}$.

$$
\langle f, g \rangle
= \int_a^b f(x) \cdot g(x) \,\mathrm{d}x
= \int_a^b g(x) \cdot f(x) \,\mathrm{d}x
= \langle g, f \rangle
$$

2. Folgt aus der Linearität des Integrals:

$$
\langle f + a \cdot g, h \rangle
= \int_a^b \left(f(x) + a \cdot g(x)\right) \cdot h(x) \,\mathrm{d}x
= \int_a^b f(x) \cdot h(x) \,\mathrm{d}x + a \int_a^b g(x) \cdot h(x) \,\mathrm{d}x
= \langle f, h \rangle + a \langle g, h \rangle
$$

Übrigens gilt die Linearität auch für die rechte Seite des Skalarprodukts:

$$ \langle f, g + a \cdot h \rangle = \langle f, g \rangle + a \langle f, h \rangle $$

Das Skalarprodukt ist also linear in beiden Argumenten, was man auch als *bilinear* bezeichnet.

#### 3.2) Skalarprodukt von Sinus und Cosinus

Die beiden Funktionen

$$ f_n(t) = \sin\left(n \omega_0 t\right) \qquad\text{und}\qquad g_n(t) = \cos\left(n \omega_0 t\right) $$

beschreiben harmonische Schwingungen mit der Kreisfrequenz $\omega_n = n \omega_0$, wobei $n\in\mathbb{N}$ eine natürliche Zahl grösser Null und $\omega_0\in\mathbb{R}_{>0}$ eine positive reelle Konstante ist. Die Kreisfrequenz $\omega_0$ ist die Grundschwingung und für $n>1$ ist $\omega_n$ eine Oberschwingung.

**Aufgabe:** Zeige, dass die beiden Funktionen bezüglich des Skalarprodukts über eine Periode der Grundschwingung, also $\left[0,\frac{2\pi}{\omega_0}\right]$, orthogonal sind.

**Hinweis:** Verwende das [Additionstheorem](https://de.wikipedia.org/wiki/Formelsammlung_Trigonometrie#Additionstheoreme)

$$ \sin(\alpha)\cos(\beta) + \cos(\alpha)\sin(\beta) = \sin(\alpha+\beta) $$

mit $\alpha=\beta$, um den Integranden zu vereinfachen, bevor zu die Stammfunktion bestimmst.

**Lösung:**

\begin{align*}
\langle f_n, g_n \rangle
&= \int_0^{2\pi/\omega_0} \sin\left(n \omega_0 t\right) \cdot \cos\left(n \omega_0 t\right) \,\mathrm{d}t \\
&= \frac{1}{2} \int_0^{2\pi/\omega_0} \sin\left(2 n \omega_0 t\right) \,\mathrm{d}t \\
&= \frac{1}{2} \bigg[ -\frac{1}{2 n \omega_0} \cos\left(2 n \omega_0 t\right) \bigg]_0^{2\pi/\omega_0} \\
&= - \frac{1}{4 n \omega_0} \Big( \cos(4 \pi n) - \cos(0) \Big) \\
&= - \frac{1}{4 n \omega_0} \left( 1 - 1 \right) \\
&= 0
\end{align*}

### Aufgabe 4: Faltung von Funktionen

Berechne die Faltung $(f*g)(t)$ von

1. der Rechteckfunktion, $f(t)=\begin{cases} 1 & \text{für } t\in[0,1] \\ 0 & \text{sonst} \end{cases}$, und der Funktion $g(t)=\sqrt{t}$.

2. $f(t)=e^{-t^2}$ und $g(t)=\sin(t)$ mithilfe von
   - a) Wolfram|Alpha analytisch
   - b) SciPy numerisch
   - c) und vergleiche die beiden im Bereich $t\in[-2\pi,2\pi]$. 

**Lösungen**

1. 

\begin{align*}
\int_{-\infty}^{\infty} f(\tau) g(t-\tau) \,\mathrm{d}\tau
&= \int_0^1 (t - \tau)^{1/2} \mathrm{d}\tau \\
&= \bigg[ - \frac{2}{3} ( t - \tau )^{3/2} \bigg]_0^1 \\
&= - \frac{2}{3} (t - 1)^{3/2} + \frac{2}{3} t^{3/2} \\
&= \frac{2}{3} \left( t^{3/2} - (t - 1)^{3/2} \right)
\end{align*}

2.a)

Das Integral kann mit [Wolfram|Alpha: integrate exp(-x^2)sin(t-x) dx from -inf to inf](https://www.wolframalpha.com/input?i=integrate+exp%28-x%5E2%29sin%28t-x%29+dx+from+-inf+to+inf) berechnet werden.

\begin{align*}
\int_{-\infty}^{\infty} f(\tau) g(t-\tau) \,\mathrm{d}\tau
&= \int_{-\infty}^{\infty} e^{-\tau^2}\sin(t - \tau) \,\mathrm{d}\tau \\
&= \frac{\sqrt{\pi}}{\sqrt[4]{e}} \cdot \sin(t) \\
&\approx 1.380 \cdot \sin(t)
\end{align*}

```python
# 2.b)
from scipy.integrate import quad

f = lambda t: np.exp(-(t**2))
g = np.sin

conf_fg = np.vectorize(
    lambda t: quad(lambda tau: f(tau) * g(t - tau), -np.inf, np.inf)[0]
)

t = np.linspace(-2 * np.pi, 2 * np.pi, 100)
cn = conf_fg(t)
ca = np.sqrt(np.pi) * np.exp(-1 / 4) * np.sin(t)

plt.plot(t, ca, label=r"$f \ast g$ (analytisch)")
plt.plot(t, cn, ".", label=r"$f \ast g$ (numerisch)")
plt.xlabel(r"$t$")
plt.ylabel(r"$f \ast g$")
plt.legend()
plt.show()
```

### Aufgabe 5: Glättung von Daten mit der Faltung

#### 5.2) Faltung mit einer Rechteckfunktion

Nun soll die Herzfrequenz geglättet werden. Dazu wird die Faltung der Herzfrequenz mit einer Rechteckfunktion durchgeführt. Die Breite der Rechteckfunktion soll als Parameter `width=30` definiert werden, so dass sie später einfach verändert werden kann.

**a)** Benutze die Funktion [`np.convolve`](https://numpy.org/doc/stable/reference/generated/numpy.convolve.html) mit `mode="valid"`, um die diskrete Faltung zu berechnen. Siehe Beispiel in der Vertiefung. Stelle die Messdaten und die geglätteten Daten dar.

```python
# 5.2.a)

width = 30
window = np.full(width, 1 / width)
hr_smooth = np.convolve(hr_data, window, mode="valid")
t_smooth = t_data[width // 2 : -width // 2 + 1]

plt.plot(t_data, hr_data, label="Daten")
# plt.xlim(t_data.min(), t_data.max())
plt.xlim(500, 1000)
plt.ylim(150, 170)
plt.plot(t_smooth, hr_smooth, label="Gleitender Mittelwert")
plt.xlabel("Zeit [s]")
plt.ylabel("Herzfrequenz [1/min]")
plt.legend()
plt.grid()
plt.show()
```

**b)** Berechne nun die Faltung mittels Integration. Erstelle dazu zuerst eine Interpolationsfunktion der Herzfrequenz und die Rechteckfunktion. Definiere dann eine Funktion `hr_convolution(t)`, welche die Faltung der Rechteckfunktion mit der Herzfrequenz-Interpolation berechnet. Stelle diese Funktion auch noch zusammen mit den Plots aus Teil a) da. (Die beiden Resultate sollten übereinstimmen.)

```python
from scipy.interpolate import make_interp_spline

half_width = width / 2

rect = lambda t: np.where(np.abs(t) <= half_width, 1 / width, 0)

hr_spline = make_interp_spline(t_data, hr_data)


def hr_convolution(t, limit=100):
    integrand = lambda tau: rect(tau) * hr_spline(t - tau)
    return quad(integrand, -half_width, half_width, limit=limit)[0]


hr_smooth_2 = np.vectorize(hr_convolution)(t_smooth)

plt.plot(t_data, hr_data, label="Herzfrequenz")
plt.plot(t_smooth, hr_smooth, label="Gleitender Mittelwert (np.convolve)")
plt.plot(t_smooth, hr_smooth_2, "--", label="Gleitender Mittelwert (integral)")
# plt.xlim(t_data.min(), t_data.max())
plt.xlim(500, 1000)
plt.ylim(150, 170)
plt.xlabel("Zeit [s]")
plt.ylabel("Herzfrequenz [1/min]")
plt.legend()
plt.grid()
plt.show()
```
